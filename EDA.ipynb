```python
# Panji Malik Ibrahim_202431202

print("Gender:", df['gender'].unique())
print("Race/Ethnicity:", df['race/ethnicity'].unique())
print("Parental Education:", df['parental level of education'].unique())
print("Lunch:", df['lunch'].unique())
print("Test Prep Course:", df['test preparation course'].unique())
```

```python
# Panji Malik Ibrahim_202431202

# Imputasi kolom angka dengan mean
for col in df.select_dtypes(include=['int64', 'float64']).columns:
    df[col] = df[col].fillna(df[col].mean())

# Imputasi kolom kategorik dengan modus
for col in df.select_dtypes(include=['object']).columns:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
# Panji Malik Ibrahim_202431202

df.duplicated().sum()
```

```python
# Panji Malik Ibrahim_202431202

df = df.drop_duplicates()
df.info()
```

```python
# Panji Malik Ibrahim_202431202

df['parental level of education'].value_counts().plot(kind='bar')
plt.title('Jumlah Siswa Berdasarkan Tingkat Pendidikan Orang Tua')
plt.xlabel('Tingkat Pendidikan Orang Tua')
plt.ylabel('Jumlah Siswa')
plt.show()
```
