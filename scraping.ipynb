```python
# Panji Malik Ibrahim_202431202

import requests
import csv
from bs4 import BeautifulSoup
import pandas as pd

print("Library berhasil di-import!")
```

```python
# Panji Malik Ibrahim_202431202

# Ambil halaman web utama & cek status code
url = "http://books.toscrape.com/"
response = requests.get(url)

print("Status Code:", response.status_code)

if response.status_code == 200:
    print("Berhasil mengakses web target (HTTP 200 OK)!")
else:
    print(f"Gagal mengakses web. Status code: {response.status_code}")
```

```python
# Panji Malik Ibrahim_202431202

# Parsing HTML menggunakan BeautifulSoup
soup = BeautifulSoup(response.text, 'html.parser')

print("Parsing HTML berhasil!")
print("Judul Halaman:", soup.title.string.strip())
```

```python
# Panji Malik Ibrahim_202431202

# Scraping data buku dari books.toscrape.com
base_url = "http://books.toscrape.com/catalogue/page-{}.html"
data_buku = []

# Loop halaman 1 sampai 15
for page in range(1, 16):
    url = base_url.format(page)
    res = requests.get(url)

    if res.status_code != 200:
        print(f"Gagal mengakses halaman {page}")
        break

    soup_page = BeautifulSoup(res.text, 'html.parser')
    books = soup_page.find_all('article', class_='product_pod')

    for item in books:
        # 1. Judul Buku
        title = item.h3.a['title']

        # 2. Harga (£)
        price = item.find('p', class_='price_color').text

        # 3. Rating Bintang (One, Two, Three, Four, Five)
        rating_class = item.find('p', class_='star-rating')['class'][1]

        data_buku.append({
            'Judul_Buku': title,
            'Harga_GBP': price,
            'Rating_Bintang': rating_class
        })

print(f"Total data berhasil di-scrape: {len(data_buku)} data")
```

```python
# Panji Malik Ibrahim_202431202

# Pembersihan Data

# 1. Konversi ke DataFrame & Hapus Duplikat
df = pd.DataFrame(data_buku)
df = df.drop_duplicates()

# 2. Bersihkan kolom Harga
df['Harga_GBP'] = df['Harga_GBP'].str.replace('£', '', regex=False).str.replace('Â', '', regex=False).astype(float)

# 3. Ubah Rating Teks ke Angka (1-5)
rating_map = {'One': 1, 'Two': 2, 'Three': 3, 'Four': 4, 'Five': 5}
df['Rating_Angka'] = df['Rating_Bintang'].map(rating_map)

# 4. Ambil 3 kolom
df_clean = df[['Judul_Buku', 'Harga_GBP', 'Rating_Angka']]

df_clean.to_excel('hasil_scraping_books.xlsx', index=False)

print("Langkah 5 Selesai! Berkas 'hasil_scraping_books.xlsx' berhasil dibuat.")
print(f"Total data bersih: {len(df_clean)} baris.")
```
